```python
#loading in our csv files
import pandas as pd
import os

csv_directory = "data/raw_data"

csv_files = [file for file in os.listdir(csv_directory) if file.endswith(".csv")]
```

```python
#creating an empty dataframe to merge the csv files
merged_census_data = pd.DataFrame()
```

```python
#looping through CSV files
for csv_file in csv_files:
    csv_path = os.path.join(csv_directory, csv_file) #creating a consistent path
    df_csv = pd.read_csv(csv_path, low_memory=False) #reading each file
    merged_census_data = pd.concat([merged_census_data, df_csv], axis=1) #concatenating all columns, duplicate oa_codes will have to be eliminated
```

```python
#saving the new merged data
merged_census_data.to_csv("data/raw_data/merged_census_data.csv", index=False)
```

```python
#exploring the new merged dataframe
merged_census_data.head(3)
```

```python
#we now make a filter to only work with Brighton and Hove output areas

#first, reading in our shapefile of all output areas for the UK
import geopandas as gpd

shp_path = "data/raw_data/OA_2021_EW_BGC_V2.shp"
gdf = gpd.read_file(shp_path)
gdf.head()
```

```python
#searching for Brighton and Hove
brighton_hove_rows = gdf[gdf["LSOA21NM"].str.contains("Brighton|Hove", case=False, na=False)]
brighton_hove_rows.head()
```

```python
#saving Brighton and Hove output areas as a shapefile
brighton_hove_rows.to_file("data/raw_data/brighton_hove_output_areas.shp")
```

```python
#loading the filtered shapefile
filtered_shapefile = "data/raw_data/brighton_hove_output_areas.shp"
gdf = gpd.read_file(filtered_shapefile)
```
